```python
# ========================================================
# 06_model_comparison.ipynb
# Porównanie AE, Isolation Forest i LSTM AE po rozszerzeniu datasetu
# ========================================================

import pandas as pd
import os

print("=== FINALNE PORÓWNANIE MODELI v2 ===")

ae = pd.read_csv("../results/final_ae_results.csv")
iforest = pd.read_csv("../results/isolation_forest_results.csv")
lstm = pd.read_csv("../results/lstm_ae_window5_results.csv")

print("AE shape:", ae.shape)
print("IF shape:", iforest.shape)
print("LSTM shape:", lstm.shape)

comparison = ae.merge(
    iforest,
    on="scenario",
    how="left"
).merge(
    lstm,
    on="scenario",
    how="left"
)

comparison["AE - IF (pkt proc.)"] = (
    comparison["Autoencoder DR (%)"] - comparison["Isolation Forest DR (%)"]
).round(2)

comparison["AE - LSTM (pkt proc.)"] = (
    comparison["Autoencoder DR (%)"] - comparison["LSTM AE DR (%)"]
).round(2)

comparison["LSTM - IF (pkt proc.)"] = (
    comparison["LSTM AE DR (%)"] - comparison["Isolation Forest DR (%)"]
).round(2)

cols = [
    "scenario",

    "AE_detected",
    "AE_total",
    "Autoencoder DR (%)",

    "IF_detected",
    "IF_total",
    "Isolation Forest DR (%)",

    "LSTM_detected",
    "LSTM_total_windows",
    "LSTM AE DR (%)",

    "AE - IF (pkt proc.)",
    "AE - LSTM (pkt proc.)",
    "LSTM - IF (pkt proc.)",
]

comparison_final = comparison[cols].copy()

print("\n=== PORÓWNANIE MODELI ===")
print(comparison_final.to_string(index=False))

os.makedirs("../results", exist_ok=True)
comparison_final.to_csv("../results/final_comparison.csv", index=False)

print("\nZapisano: ../results/final_comparison.csv")
```
