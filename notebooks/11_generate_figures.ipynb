```python
# ========================================================
# 11_generate_figures.ipynb
# ========================================================

import pandas as pd
import matplotlib.pyplot as plt
import os

# Output directory for generated figures
os.makedirs("../results/figures", exist_ok=True)

print("=== GENERATING PROJECT FIGURES ===")

# ========================================================
# 1. Anomaly detection rate comparison
# ========================================================

df_results = pd.read_csv("../results/final_comparison_table.csv")

scenario_labels = {
    "normal (FPR)": "Normal (FPR)",
    "guloader": "Guloader",
    "scanning": "Scanning",
    "njrat": "NjRAT",
    "kongtuke1": "KongTuke 1",
    "kongtuke2": "KongTuke 2",
    "remcos": "Remcos",
    "xloader": "XLoader",
    "xworm": "XWorm",
    "phantomstealer": "PhantomStealer"
}

labels = [
    scenario_labels.get(scenario, scenario)
    for scenario in df_results["Scenariusz"]
]

plt.figure(figsize=(13, 6))

x = range(len(df_results))
bar_width = 0.25

plt.bar(
    [i - bar_width for i in x],
    df_results["Autoencoder DR (%)"],
    width=bar_width,
    label="Autoencoder"
)

plt.bar(
    x,
    df_results["Isolation Forest DR (%)"],
    width=bar_width,
    label="Isolation Forest"
)

plt.bar(
    [i + bar_width for i in x],
    df_results["LSTM AE DR (%)"],
    width=bar_width,
    label="LSTM Autoencoder"
)

plt.xticks(x, labels, rotation=45, ha="right")
plt.ylabel("Detection rate [%]")
plt.xlabel("Traffic scenario")
plt.title("Anomaly Detection Rate by Model and Traffic Scenario")
plt.ylim(0, 110)
plt.grid(axis="y", alpha=0.3)
plt.legend()
plt.tight_layout()

plt.savefig(
    "../results/figures/detection_rate_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: ../results/figures/detection_rate_comparison.png")


# ========================================================
# 2. Anomaly-family classification confusion matrix
# ========================================================

df_cm = pd.read_csv("../results/family_classification_confusion_matrix.csv", index_col=0)

plt.figure(figsize=(10, 8))

plt.imshow(df_cm.values)
plt.title("Anomaly Family Classification Confusion Matrix")
plt.colorbar()

plt.xticks(range(len(df_cm.columns)), df_cm.columns, rotation=45, ha="right")
plt.yticks(range(len(df_cm.index)), df_cm.index)

plt.xlabel("Predicted class")
plt.ylabel("True class")

for i in range(len(df_cm.index)):
    for j in range(len(df_cm.columns)):
        plt.text(j, i, df_cm.values[i, j], ha="center", va="center")

plt.tight_layout()

plt.savefig("../results/figures/family_classification_confusion_matrix.png", dpi=300)
plt.show()

print("Saved: ../results/figures/family_classification_confusion_matrix.png")


print("\nDone. Figures saved to: ../results/figures/")
```
