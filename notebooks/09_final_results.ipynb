```python
# ========================================================
# 09_final_results.ipynb
# Final model comparison table
# ========================================================

import pandas as pd
import os

print("=== FINAL MODEL COMPARISON TABLE ===")

comparison = pd.read_csv("../results/final_comparison.csv")

df_final = comparison.rename(columns={
    "scenario": "Scenariusz",
    "AE_detected": "AE wykryte",
    "AE_total": "AE liczba flowów",
    "IF_detected": "IF wykryte",
    "IF_total": "IF liczba flowów",
    "LSTM_detected": "LSTM wykryte",
    "LSTM_total_windows": "LSTM liczba okien",
})

cols = [
    "Scenariusz",

    "AE wykryte",
    "AE liczba flowów",
    "Autoencoder DR (%)",

    "IF wykryte",
    "IF liczba flowów",
    "Isolation Forest DR (%)",

    "LSTM wykryte",
    "LSTM liczba okien",
    "LSTM AE DR (%)",

    "AE - IF (pkt proc.)",
    "AE - LSTM (pkt proc.)",
    "LSTM - IF (pkt proc.)",
]

df_final = df_final[cols].copy()

# Zaokrąglenia
for c in [
    "Autoencoder DR (%)",
    "Isolation Forest DR (%)",
    "LSTM AE DR (%)",
    "AE - IF (pkt proc.)",
    "AE - LSTM (pkt proc.)",
    "LSTM - IF (pkt proc.)",
]:
    df_final[c] = df_final[c].round(2)

print("\n=== FINAL RESULTS ===")
print(df_final.to_string(index=False))

os.makedirs("../results", exist_ok=True)
df_final.to_csv("../results/final_comparison_table.csv", index=False)

print("\nSaved: ../results/final_comparison_table.csv")
```
